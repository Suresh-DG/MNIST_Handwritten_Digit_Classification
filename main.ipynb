```python
#importing libraries
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

```python
#loading Data
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
#Data Preprocessing
x_train = x_train/255
x_test = x_test/255
```

```python
x_train[0]
```

```python
x_train_flattend = x_train.reshape(len(x_train), 28*28)
x_test_flattend = x_test.reshape(len(x_test), 28*28)
x_train_flattend.shape
```

```python
x_test_flattend.shape
```

```python
x_test_flattend[0]
```

```python
#Model_1 - No Hidden Layer
model = keras.Sequential([
    keras.layers.Dense(10, input_shape=(784,),activation='sigmoid')
])
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model.fit(x_train_flattend, y_train, epochs=5)
```

```python
model.evaluate(x_test_flattend, y_test)
```

```python
plt.matshow(x_test[0])
```

```python
y_predicted = model.predict(x_test_flattend)
y_predicted[0]
```

```python
np.argmax(y_predicted[0])
```

```python
y_predicted_labels = [np.argmax(i) for i in y_predicted]
y_predicted_labels[:5]
```

```python
y_test[:5]
```

```python
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels)
cm
```

```python
import seaborn as sn
plt.figure(figsize=(10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
#Model_2 - with Hidden layer
model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,),activation='relu'),
    keras.layers.Dense(10, activation='sigmoid')
])
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
model.fit(x_train_flattend, y_train, epochs=5)
```

```python
model.evaluate(x_test_flattend,y_test)
```

```python
y_predicted = model.predict(x_test_flattend)
y_predicted_labels = [np.argmax(i) for i in y_predicted]
cm = tf.math.confusion_matrix(labels=y_test,predictions=y_predicted_labels)
plt.figure(figsize=(10,7))
sn.heatmap(cm, annot=True, fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
#Baseline with a Validation Split
baseline_model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

baseline_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history = baseline_model.fit(
    x_train_flattend, y_train,
    epochs=10,
    validation_split=0.1,
    verbose=1
)

baseline_test_loss, baseline_test_accuracy = baseline_model.evaluate(x_test_flattend, y_test, verbose=0)
print(f"Baseline Test Accuracy: {baseline_test_accuracy:.4f}  |  Test Loss: {baseline_test_loss:.4f}")
```

```python
#Visulaisation
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].plot(history.history['accuracy'], label='Training Accuracy', marker='o')
axes[0].plot(history.history['val_accuracy'], label='Validation Accuracy', marker='o')
axes[0].set_title('Accuracy over Epochs')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Accuracy')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(history.history['loss'], label='Training Loss', marker='o')
axes[1].plot(history.history['val_loss'], label='Validation Loss', marker='o')
axes[1].set_title('Loss over Epochs')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig('training_curves.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
#Testing model on five handwritten images
np.random.seed(7)
sample_indices = np.random.choice(len(x_test_flattend), 5, replace=False)

sample_predictions = baseline_model.predict(x_test_flattend[sample_indices], verbose=0)
predicted_labels_sample = np.argmax(sample_predictions, axis=1)
actual_labels_sample = y_test[sample_indices]

fig, axes = plt.subplots(1, 5, figsize=(14, 3))
for i, idx in enumerate(sample_indices):
    axes[i].imshow(x_test[idx], cmap='gray')
    correct = predicted_labels_sample[i] == actual_labels_sample[i]
    color = 'green' if correct else 'red'
    axes[i].set_title(f"Actual: {actual_labels_sample[i]}\nPredicted: {predicted_labels_sample[i]}", color=color)
    axes[i].axis('off')
plt.suptitle("Model Predictions on 5 Test Images (Actual vs Predicted)")
plt.tight_layout()
plt.savefig('predictions_sample.png', dpi=150, bbox_inches='tight')
plt.show()

for idx, a, p in zip(sample_indices, actual_labels_sample, predicted_labels_sample):
    print(f"Image index {idx}: Actual = {a}, Predicted = {p}  -> {'Correct' if a == p else 'Incorrect'}")
```

```python
#Experiment-Dropout
experiment_model = keras.Sequential([
    keras.layers.Dense(100, input_shape=(784,), activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(10, activation='softmax')
])

experiment_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_experiment = experiment_model.fit(
    x_train_flattend, y_train,
    epochs=10,
    validation_split=0.1,
    verbose=1
)

exp_test_loss, exp_test_accuracy = experiment_model.evaluate(x_test_flattend, y_test, verbose=0)
print(f"Experiment (Dropout) Test Accuracy: {exp_test_accuracy:.4f}  |  Test Loss: {exp_test_loss:.4f}")
```

```python
#Comparsion b/w Baseline & Dropout
print(f"{'Model':<28}{'Test Accuracy':<18}{'Test Loss':<12}")
print("-" * 58)
print(f"{'Baseline (no dropout)':<28}{baseline_test_accuracy:<18.4f}{baseline_test_loss:<12.4f}")
print(f"{'Experiment (dropout)':<28}{exp_test_accuracy:<18.4f}{exp_test_loss:<12.4f}")

plt.figure(figsize=(8, 4.5))
plt.plot(history.history['val_accuracy'], label='Baseline - Val Accuracy', marker='o')
plt.plot(history_experiment.history['val_accuracy'], label='Dropout - Val Accuracy', marker='s')
plt.title('Baseline vs Dropout: Validation Accuracy Comparison')
plt.xlabel('Epoch')
plt.ylabel('Validation Accuracy')
plt.legend()
plt.grid(alpha=0.3)
plt.savefig('baseline_vs_experiment.png', dpi=150, bbox_inches='tight')
plt.show()
```
